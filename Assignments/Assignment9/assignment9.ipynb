1. Draw a histogram showing distribution of heights across species for penguin dataset
Create a stack plot and step plot- – Write down your interpretations

```python
import seaborn as sns
import matplotlib.pylab as plt
```

```python
sns.get_dataset_names()
```

```python
df = sns.load_dataset('penguins')
df
```

```python
sns.histplot(df, x = 'flipper_length_mm', hue='species')
```

```python
sns.histplot(df, x = 'flipper_length_mm', multiple='stack', hue = 'species')
```

```python
sns.histplot(df, x = 'flipper_length_mm',  hue = 'species', element='step')
```

1. Gentoo penguins have the longest flipper lengths (around 215-230mm)
2. Chinstrap penguins have medium flipper lengths (around 190-200mm)
3. Adelie penguins have the shortest flipper lengths (around 180-190mm)

From the histogram plots, we can observe that:
1. Gentoo penguins consistently show the longest flipper lengths among all species
2. Chinstrap penguins have intermediate flipper lengths
3. Adelie penguins have the shortest flipper lengths
4. The stacked plot shows clear separation between species' size distributions
5. The step plot highlights the distinct ranges of flipper lengths for each species

Q2. Draw a bivariant KDE plot to show relationship between horsepower and displacement [use mpg dataset] – Write down your interpretations 

```python
mpg = sns.load_dataset('mpg')
mpg
```

```python
# Create a bivariate KDE plot
sns.kdeplot(data=mpg, x='horsepower', y='displacement', cmap='viridis', fill=True,levels=10,thresh=0)
plt.title('Bivariate KDE Plot: Horsepower vs Displacement')
plt.xlabel('Horsepower')
plt.ylabel('Displacement')
```

* If there's a positive relationship, you'll see the darker colors running from
bottom-left to top-right
* Higher density (darker areas) shows where most cars cluster
* The contour lines (like topographic maps) show regions of equal density

3. Draw a correlation heatmap for diamonds dataset, add annotation, change colour pallet and write your interpretation 

```python
diamonds = sns.load_dataset('diamonds')
diamonds
```

```python
diamonds = diamonds.loc[:,'depth':]
diamonds
```

```python
sns.heatmap(diamonds.corr(), annot=True, cmap='crest' )
```

Diamond Correlation Heatmap Interpretation:

1. Positive Correlations:
   - There is a strong positive correlation between carat and price
   - As diamond size (carat) increases, the price tends to increase proportionally
   - The x, y, z dimensions show strong correlations with each other

2. Color Interpretation:
   - Darker blue regions indicate stronger positive correlations
   - Lighter blue regions indicate weaker correlations
   - The diagonal shows perfect correlation (1.0) as expected

3. Price Relationships:
   - Price is most strongly influenced by carat weight
   - Physical dimensions (x, y, z) have secondary influence on price
   - Depth and table have relatively weaker correlations with price
