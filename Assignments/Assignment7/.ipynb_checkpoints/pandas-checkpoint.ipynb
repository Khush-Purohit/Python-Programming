```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame(data = [['aa', 18, 'CSE', 85]], 
                  index = ['student1'],
                  columns = ['name', 'age', 'department', 'marks'])
df
```

```python
data = {
    'Name': ['aa','bb','cc'],
    'age' : [18,21,22],
    'department': ['cse','ece','eee'],
    'marks' : [85, 75,65]
}
df = pd.DataFrame(data)
df
```

```python
maximum = (df['marks'].max())
minimum = (df['marks'].min())
avg = (df['marks'].mean())
print(f'avg is {avg}, max is {maximum}, min is {minimum}')
```
