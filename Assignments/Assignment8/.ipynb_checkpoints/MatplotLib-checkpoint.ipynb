```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("Population.csv")
df
```

```python
pop = df['Population']
urb = df['Urban Population']
rural = pop - urb
rural
```

```python

plt.plot(rural,urb, label = "Urban pop. vs rural pop.")
plt.xlabel("Urban population")
plt.ylabel("Rural population")
plt.legend()
plt.show()
```

```python
# print(rural)
# print(urb)
plt.bar(rural,urb,color ='blue')
plt.title('Bar chart of uraban vs rural')
plt.xlabel("Urban Population")
plt.ylabel("Rural Population")
plt.show()
```
