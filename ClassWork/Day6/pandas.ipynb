```python
head = pd.Series({'usa':'nyc', 'india': 'pune', 'uk':'london', 'france':'paris'})
head
```

```python
# combining series data- should have same indices
df = pd.DataFrame({'Head Office': head, 'Employee Count':count})
df
```

```python
#===== end index is also included here!!=====
df.loc['india':'france']
```

```python
df.loc['india':'france', 'Head Office	' : 'Employee Count']
```

```python
df.iloc[1:3, [0,1]]
#loc should have only labeled index, only if index is not given, then it takes integer as an index
```
