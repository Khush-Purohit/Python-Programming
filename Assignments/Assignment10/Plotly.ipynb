2. Draw a Histogram showing votes secured by 3 candidates - Coderre, Joly, Bergeron
Use semantics -bar mode, color, facet_row to improve your chart. Create a cumulative graph as well. [use election dataset]

```python
election = px.data.election()
election.head()
```

```python
votes = election[['Coderre', 'Bergeron','Joly']]
votes.head(5)
```

```python
# votes_melted = votes.melt(var_name=[['Coderre','Bergeron','Joly']], value_name='total')
votes_melt = votes.melt(var_name='Candidate', value_name='Votes')
px.histogram(votes_melt, x='Votes', color='Candidate')

# fig = px.histogram(votes)
# fig.show()
```

```python
# First properly melt the data
votes_melted = votes.melt(var_name='Candidate', value_name='Votes')
# Create histogram with multiple traces
fig = px.histogram(votes_melted, 
                  x='Votes',
                  color='Candidate',
                  barmode='overlay',  # Shows overlapping distributions
                  facet_row='Candidate',  # Creates separate rows for each candidate
                  title='Vote Distribution by Candidate')

# Add cumulative histogram
fig_cumulative = px.histogram(votes_melted,
                            x='Votes',
                            color='Candidate',
                            cumulative=True,
                            title='Cumulative Vote Distribution')

# Show both plots
fig.show()
fig_cumulative.show()
```
