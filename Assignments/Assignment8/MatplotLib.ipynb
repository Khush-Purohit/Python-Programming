1. Draw line & bar charts to show urban and rural population of India over the years using population.csv [rural population is the difference between total population and urban population][For bar chart consider 8 yrs data]

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("Population.csv")
df
```

```python
pop = df['Population']
urb = df['Urban Population']
rural = pop - urb
rural
```

```python

plt.plot(rural,urb, label = "Urban pop. vs rural pop.")
plt.xlabel("Urban population")
plt.ylabel("Rural population")
plt.legend()
plt.show()
```

```python
# print(rural)
# print(urb)
# print(f'rural: {rural}')
# print(f'urb: {urb}')
rural = rural[:9]
urb = urb[:9]
print(f'rural: {rural}')
print(f'urb: {urb}')
plt.bar(rural,urb,color ='blue')
plt.title('Bar chart of uraban vs rural')
plt.xlabel("Urban Population")
plt.ylabel("Rural Population")
plt.show()
```

3. Using a pie chart show distribution of top six selling cars [Create data as required]  

```python
cars = ['Maruti Swift', 'Hyundai Creta', 'Tata Nexon', 'Mahindra XUV700', 'Toyota Fortuner', 'Honda City','Maruti Baleno', 'Tata Punch', 'Hyundai Venue', 'Kia Seltos', 'MG Hector']
sales = [25000, 18000,  6500, 6000,7500, 12000, 10000,7000, 6500, 6000, 5500]

car = pd.DataFrame(
    {
        'carModel': cars,
        'sales':sales
    }
)
car
```

```python
car = car.sort_values(by='sales', ascending=False).head(6)
car
```

```python
plt.pie(car['sales'], labels=car['carModel'],autopct='%1.1f%%',explode=[0.1,0,0,0,0,0],shadow=True)
plt.show()
```
